**Q) Calculate weekly paycheck of a person paid hourly, at a rate of $15/hour paying taxes(12%)**

```python
hourlyRate = 15       # dollars per hour
hoursWorked = 40      # hours per week
taxRate = 0.12        # 12% tax

grossPay = hourlyRate * hoursWorked
taxAmount = grossPay * taxRate
netPay = grossPay - taxAmount

print("Gross Pay: $",grossPay)
print("Tax Deducted: $",taxAmount)
print("Net Paycheck: $",netPay)
```

```python
def calculateGrossPay(hourlyRate, hoursWorked):
    return hourlyRate * hoursWorked

def calculateTax(grossPay, taxRate):
    return grossPay * taxRate

def calculateNetPay(hourlyRate, hoursWorked, taxRate):
    grossPay = calculateGrossPay(hourlyRate, hoursWorked)
    taxAmount = calculateTax(grossPay, taxRate)
    netPay = grossPay - taxAmount
    return grossPay, taxAmount, netPay

hourlyRate = 15        # dollars/hour
hoursWorked = 40       # hours/week
taxRate = 0.12         # 12%

grossPay, taxAmount, netPay = calculateNetPay(hourlyRate, hoursWorked, taxRate)

print(f"Gross Pay: ${grossPay}")
print(f"Tax Deducted: ${taxAmount}")
print(f"Net Paycheck: ${netPay}")
```

```python
"""
Weekly Paycheck Calculator

This script calculates the weekly paycheck for an hourly worker, accounting for tax deductions.
"""

def calculateGrossPay(hourlyRate: float, hoursWorked: float) -> float:
    """
    Calculate the gross pay before taxes.

    Parameters:
        hourlyRate (float): Hourly wage.
        hoursWorked (float): Number of hours worked in a week.

    Returns:
        float: Gross pay.
    """
    return hourlyRate * hoursWorked

def calculateTax(grossPay: float, taxRate: float) -> float:
    """
    Calculate the tax based on gross pay.

    Parameters:
        grossPay (float): Total pay before tax.
        taxRate (float): Tax rate as a decimal (e.g., 0.12 for 12%).

    Returns:
        float: Tax amount.
    """
    return grossPay * taxRate

def calculateNetPay(hourlyRate: float, hoursWorked: float, taxRate: float) -> tuple:
    """
    Calculate gross pay, tax amount, and net pay.

    Parameters:
        hourlyRate (float): Hourly wage.
        hoursWorked (float): Number of hours worked.
        taxRate (float): Tax rate as a decimal.

    Returns:
        tuple: (grossPay, taxAmount, netPay)
    """
    grossPay = calculateGrossPay(hourlyRate, hoursWorked)
    taxAmount = calculateTax(grossPay, taxRate)
    netPay = grossPay - taxAmount
    return grossPay, taxAmount, netPay

def getFloatInput(prompt: str) -> float:
    """
    Prompt the user for a float input and validate it.

    Parameters:
        prompt (str): The message to display for input.

    Returns:
        float: Validated float input from the user.
    """
    while True:
        try:
            value = float(input(prompt))
            if value < 0:
                raise ValueError("Value must be non-negative.")
            return value
        except ValueError as e:
            print(f"Invalid input: {e}")

def displayPayDetails(grossPay: float, taxAmount: float, netPay: float):
    """
    Display formatted pay details.

    Parameters:
        grossPay (float): Gross pay before tax.
        taxAmount (float): Tax amount deducted.
        netPay (float): Final take-home pay.
    """
    print("\n----- Paycheck Summary -----")
    print(f"Gross Pay     : ${grossPay:.2f}")
    print(f"Tax Deducted  : ${taxAmount:.2f}")
    print(f"Net Paycheck  : ${netPay:.2f}")

def main():
    print("=== Weekly Paycheck Calculator ===")

    hourlyRate = getFloatInput("Enter hourly rate ($): ")
    hoursWorked = getFloatInput("Enter hours worked: ")
    taxRate = getFloatInput("Enter tax rate (e.g., 0.12 for 12%): ")

    grossPay, taxAmount, netPay = calculateNetPay(hourlyRate, hoursWorked, taxRate)
    displayPayDetails(grossPay, taxAmount, netPay)

if __name__ == "__main__":
    main()
```
