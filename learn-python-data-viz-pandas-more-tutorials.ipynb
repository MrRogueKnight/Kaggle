```python
num_years = 4
days_per_year = 365 
hours_per_day = 24
mins_per_hour = 60
secs_per_min = 60

# Calculate number of seconds in four years
total_secs = secs_per_min * mins_per_hour * hours_per_day * days_per_year * num_years
print(total_secs)

# Update to include leap years
days_per_year = 365.25 

# Calculate number of seconds in four years
total_secs = secs_per_min * mins_per_hour * hours_per_day * days_per_year * num_years
print(total_secs)

# or 

days_per_year = 365 + 1/4

# Calculate number of seconds in four years
total_secs = secs_per_min * mins_per_hour * hours_per_day * days_per_year * num_years
print(total_secs)
```

**Q) Calculate weekly paycheck of a person paid hourly, at a rate of $15/hour paying taxes(12%)**

```python
def get_pay(num_hours):
    pay_pre_tax = num_hours * 15
    pay_after_tax = pay_pre_tax * (1 - 0.12)
    return pay_after_tax
```

```python
# Calculate pay based on working 40 hours
pay_full_time = get_pay(40)
print(pay_full_time)
```

```python
pay_part_time = get_pay(32)
print(pay_part_time)
```

-> Variables defined inside the function body cannot be accessed outside of the function. For instance, the next code cell errors, because pay_after_tax only exists inside the function.

-> Variables defined inside a function (like pay_after_tax) have a local scope of that function only. 

-> Variables defined outside all functions (like pay_part_time) have a global scope and can be accessed anywhere.

**Functions with multiple arguments**

```python
def get_pay_with_more_inputs(num_hours, hourly_wage, tax_bracket):
    # Pre-tax pay
    pay_pretax = num_hours * hourly_wage
    # After-tax pay
    pay_aftertax = pay_pretax * (1 - tax_bracket)
    return pay_aftertax
```

```python
higher_pay_aftertax = get_pay_with_more_inputs(40, 24, .22)
print(higher_pay_aftertax)
```

```python
same_pay_fulltime = get_pay_with_more_inputs(40, 15, .12)
print(same_pay_fulltime)
```

```python
def calculateGrossPay(hourlyRate, hoursWorked):
    return hourlyRate * hoursWorked

def calculateTax(grossPay, taxRate):
    return grossPay * taxRate

def calculateNetPay(hourlyRate, hoursWorked, taxRate):
    grossPay = calculateGrossPay(hourlyRate, hoursWorked)
    taxAmount = calculateTax(grossPay, taxRate)
    netPay = grossPay - taxAmount
    return grossPay, taxAmount, netPay

hourlyRate = 15        # dollars/hour
hoursWorked = 40       # hours/week
taxRate = 0.12         # 12%

grossPay, taxAmount, netPay = calculateNetPay(hourlyRate, hoursWorked, taxRate)

print(f"Gross Pay: ${grossPay}")
print(f"Tax Deducted: ${taxAmount}")
print(f"Net Paycheck: ${netPay}")
```

```python
def get_pay(hourlyRate: float, hoursWorked: float, taxRate: float) -> tuple:
    """
    Calculate gross pay, tax, and net pay in one function.

    Parameters:
        hourlyRate (float): Hourly wage.
        hoursWorked (float): Hours worked in a week.
        taxRate (float): Tax rate as a decimal (e.g., 0.12 for 12%).

    Returns:
        tuple: (grossPay, taxAmount, netPay)
    """
    grossPay = hourlyRate * hoursWorked
    taxAmount = grossPay * taxRate
    netPay = grossPay - taxAmount
    return grossPay, taxAmount, netPay

def get_float_input(prompt: str) -> float:
    """Prompt the user for a valid float input."""
    while True:
        try:
            value = float(input(prompt))
            if value < 0:
                raise ValueError("Value must be non-negative.")
            return value
        except ValueError as e:
            print(f"Invalid input: {e}")

def display_summary(grossPay: float, taxAmount: float, netPay: float):
    """Display the pay details."""
    print(f"\nGross Pay     : ${grossPay:.2f}")
    print(f"Tax Deducted  : ${taxAmount:.2f}")
    print(f"Net Paycheck  : ${netPay:.2f}")

def main():
    print("=== Weekly Paycheck Calculator ===")
    hourlyRate = get_float_input("Enter hourly rate ($): ")
    hoursWorked = get_float_input("Enter hours worked: ")
    taxRate = get_float_input("Enter tax rate (e.g., 0.12 for 12%): ")
    
    grossPay, taxAmount, netPay = get_pay(hourlyRate, hoursWorked, taxRate)
    display_summary(grossPay, taxAmount, netPay)

if __name__ == "__main__":
    main()
```
